```python
import numpy as np
from scipy.io.wavfile import write, read
```

```python
for i in range(20):
    fe,x = read(f'NE/son_final_{i}.wav')
    x = x[:,0]
    x = x / np.max(np.abs(x))
    write(f'NH/son_{i}.wav', fe, x)
```
